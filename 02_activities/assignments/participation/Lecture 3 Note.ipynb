```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
# I tried ax.plot(x,y1,y2) and got the same result, is that correct?
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc="lower right", #if you put best, it means the best loc,
          frameon=True, # add frame around the legend
          fontsize=12,
          ncol=1, #specify number of columns
          shadow=False) #if put frame false, then the shadow is not making sense anymore
fig.show()
```

```python
#moving the legend outside the plot
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc="upper left",
          bbox_to_anchor= (1,1)) #attach the upper left of the legend to the (1,1) top right corner of plot
fig.show()
# loc values 'best' (Axes only) 0 'upper right' 1 'upper left'2 
# 'lower left' 3 'lower right' 4 'right' 5 'center left'6 
# 'center right' 7 'lower center' 8 'upper center' 9 'center' 10
```

```python
#text function
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
ax.text(10,95, "This value is import!",
        ha="left", #alignment of text (ha=horizontal Alignment)
        color="red",
        size=20) #(10.95 is the loc or data you want to add text)
fig.show()
```

```python
#another way for annotation
fig,ax=plt.subplots()
ax.axis([0,10,0,10]) # here create an emply set of axes
ax.text(1,5, ".Data:(1,5)", transform=ax.transData)
ax.text(0.5,0.1,".Axes:(0.5,0.1)", transform=ax.transAxes)
ax.text(0.2,0.2,".Figure:(0.2,0.2)", transform=fig.transFigure)
```

```python
#more annotation
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
ax.annotate("This is import!", xy=(10,95), xytext=(20,94),
        arrowprops=dict(facecolor="black")) #(10.95 is the loc or data you want to add text)
fig.show()
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
ax.annotate("This is import!", 
            xy=(10,95), xytext=(20,94),
            arrowprops=dict(arrowstyle="wedge", color="hotpink")) #(10.95 is the loc or data you want to add text)
fig.show()
```

```python
#remove tick marks and labels
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter()) #only remove labels
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
ax.xaxis.set_major_locator(plt.MaxNLocator(3)) #choose the best 3 major labels
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #each inerval of 5
```

```python
#if we have labels with text, it may overlape with each other, then we need rotating
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
plt.xticks(rotation=45, ha="right") #it always targets the "Current Axes" (the active subplot)
```

```python
#activity:modifying the axis title fonts
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc="lower right")
font1 = {'family':'serif', 'color':'indigo', "size":18}
plt.xlabel('Shiny New X Axis', fontdict=font1) #OR we can define the fontsize=18 here instead
```
