```python
# Activity for lecture 1
# libraries
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
 
# Dataset:
a = pd.DataFrame({ 'group' : np.repeat('A',500), 'value': np.random.normal(10, 5, 500) })
b = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(13, 1.2, 500) })
c = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(18, 1.2, 500) })
d = pd.DataFrame({ 'group' : np.repeat('C',20), 'value': np.random.normal(25, 4, 20) })
e = pd.DataFrame({ 'group' : np.repeat('D',100), 'value': np.random.uniform(12, size=100) })
df = pd.concat([a,b,c,d,e])

# plot violin chart
sns.violinplot( x='group', y='value', data=df)

# add title
plt.title("Violin plot", loc="left")

# show the graph
plt.show()
```

I choose this violin plot because it looks very interesting to me. I think it is beautiful and highly informative. It not only showing the medians and quartiles like a traditional box plot but also showing the actual distribution of the data.
Aesthetic - Mostely yes. The plot is clean and pretty. The smooth curves of the violins are visually interesting. However, there is no title and label are very simple.
Substantive - Yes. I think group b perfectly explain the benefit of this plot. If we plotted a standard bar chart, we can only see one average number in the middle, completely missing the fact that the data is split into two separate clusters.
Perceptual - Not really. We can compare the shapes and heights of (mean and distribution)the different groups but without accurate and customized axis lable, we can barely know the message under this graph.
