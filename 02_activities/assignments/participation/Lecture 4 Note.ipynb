```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1=np.arange(50)
y1=np.random.randint(0,75,50)
x2=np.array(["Liffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2=np.array([110,180,240,99,220])
```

```python
fig,(ax1,ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7,3))
ax1.scatter(x1,y1)          
ax2.bar(x2,y2)
```

```python
#Activity: Customizing our plots
fig,(ax1,ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7,3))
ax1.scatter(x1,y1, marker='*', color = 'indigo')   
font1 = {'family':'sans-serif', 'color':'blue', 'size':20}
ax1.set_title('Graph 1', fontdict=font1 )   
ax1.annotate("This is import!", 
            xy=(10,60), xytext=(20,60),
            arrowprops=dict(arrowstyle="wedge", color="hotpink"))   
ax2.bar(x2,y2, color = 'red') 
ax2.set_title('Graph 2', fontdict=font1)   
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],['ax2','ax3']],figsize=(7,4), layout="constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('A big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
# now the labels are too large to be cut off or overlap, we need to change
# we can do each one manully or try to add the layout argument
# plt.style.use('default') go back to default format
plt.show()
```

```python
#multiple viz on one axes
x=np.array(["Liffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1=np.random.randint(0,75,50)
y1=np.array([110,180,240,99,220])
y2=np.array([170,100,90,120,50])

fig,ax=plt.subplots(figsize=(7,3))
ax.bar(x,y1,color="indigo")
ax.plot(x,y2, color="red")
```

```python
#add error information
y2_sd=np.std(y2)
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2, color="red")
ax.errorbar(x,y2, yerr=y2_sd,
            fmt="none", ecolor="indigo", #if put fmt="o", there is a dot for each point
            elinewidth=4, capsize=6, capthick=4, errorevery=2) #errorevery change the frequency
```

```python
#add image to plots
from PIL import Image
import requests
from io import BytesIO

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response=requests.get(url,headers=headers)
image_file=BytesIO(response.content)
image=Image.open(image_file)
image
```

```python
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color="red")
#firstly create a container for the image
ax_image = fig.add_axes([0.1, #x coordinate
                         0.11, #y
                         0.15, #image width
                         0.35]) #image height
ax_image.imshow(image)
ax_image.axis('off') #this is no marginal line for the image
#save our viz (you need to put this code under the chunk of viz you want to save)
path = '/Users/springzhou/DSI/visualization'
filename = '/fig1a.png'
plt.savefig('/Users/springzhou/DSI/visualization/fig1a.png', dpi=300)
```
